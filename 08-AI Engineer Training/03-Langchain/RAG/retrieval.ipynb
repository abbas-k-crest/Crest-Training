```python
import os
```

```python
from langchain_community.embeddings import HuggingFaceEmbeddings

embedding = HuggingFaceEmbeddings(
    model_name="all-MiniLM-L6-v2"
)
```

## Vector Store

```python
from langchain_community.vectorstores import Chroma
```

```python
vector_store = Chroma(persist_directory= "vectordb", embedding_function=embedding)
```

```python
question = "programming languages  do Data scientist use?"
```

```python
retrieved_docs = vector_store.similarity_search(query=question, k=5)
```

```python
retrieved_docs
```

```python
for val in retrieved_docs:
    print(val.page_content)
```

## Retriever

```python
retriever = vector_store.as_retriever(search_type= 'mmr',
                                     search_kwargs = {
                                         'k': 3,
                                         'lambda': 0.7
                                     })
```

```python
retriever
```

```python
question = "What software does data scientist uses?"
```

```python
result = retriever.invoke(question)
```
