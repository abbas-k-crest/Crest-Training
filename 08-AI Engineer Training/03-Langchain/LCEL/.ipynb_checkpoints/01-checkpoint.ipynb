## Piping Prompt Model & Output Parser

```python
%load_ext dotenv
%dotenv
```

```python
import os
```

```python
from langchain_groq.chat_models import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import CommaSeparatedListOutputParser
```

```python
list_instruction = CommaSeparatedListOutputParser().get_format_instructions()
```

```python
list_instruction
```

```python
chat_template = ChatPromptTemplate.from_messages([("human", 
                                                   "I've recently adopted a {pet}. Could you suggest me three {pet} names?\n" + list_instruction)])
```

```python
chat_template
```

```python
llm = ChatGroq(
    model='openai/gpt-oss-120b',
    api_key=os.environ.get('GROQ_API_KEY'),
    max_tokens=1500
)
```

```python
list_output_paarser = CommaSeparatedListOutputParser()
```

```python
chat_template_result = chat_template.invoke({'pet': 'dog'})
```

```python
chat_result = llm.invoke(chat_template_result)
```

```python
response = list_output_paarser.invoke(chat_result)
```

```python
print(response)
```

```python
chain = chat_template | llm | list_output_paarser
```

```python
response = chain.invoke({'pet': 'dog'})
```
