## BATCH

```python
%load_ext dotenv
%dotenv
```

```python
import os
```

```python
from langchain_groq.chat_models import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import CommaSeparatedListOutputParser
```

```python
list_instruction = CommaSeparatedListOutputParser().get_format_instructions()
```

```python
list_instruction
```

```python
chat_template = ChatPromptTemplate.from_messages([("human", 
                                                   "I've recently adopted a {pet}which is a {breed}. Could you suggest me some traiing tips?\n")])
```

```python
chat_template
```

```python
llm = ChatGroq(
    model='openai/gpt-oss-120b',
    api_key=os.environ.get('GROQ_API_KEY'),
    max_tokens=1500
)
```

```python
chain = chat_template | llm
```

```python
chain.invoke({'pet': 'dog', 'breed': 'sheperd'})
```

```python
chain.batch([
    {'pet': 'dog', 'breed': 'sheperd'},
    {'pet': 'dragon', 'breed': 'night fury'}
])
```

## Stream

```python
response = chain.stream({'pet': 'dog', 'breed': 'sheperd'})
```

```python
for i in response:
    print(i.content, end="")
```
