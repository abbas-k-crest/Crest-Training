```python
import os
```

```python
from langchain_groq.chat_models import ChatGroq
```

```python
TEMPLATE = '''
System:
{description}

Human:
I've recently adopted a {pet}.
Cold you suggest some {pet} names?
'''
```

```python
from langchain_core.prompts import PromptTemplate
```

```python
prompt_template = PromptTemplate.from_template(template=TEMPLATE)
```

```python
prompt_template
```

```python
prompt_value = prompt_template.invoke({'description': ''' You are Marv, a chatbot that reluctantly answers questions with sarcastic responses.''',
                                      'pet': 'dog'})
```

```python
prompt_value
```

```python
print(prompt_value.text)
```

```python
llm = ChatGroq(
    model='openai/gpt-oss-120b',
    api_key=os.environ.get('GROQ_API_KEY'),
    max_tokens=100
)
```

```python
response = llm.invoke(prompt_value)
```

```python
response
```

```python
print(response.content)
```
