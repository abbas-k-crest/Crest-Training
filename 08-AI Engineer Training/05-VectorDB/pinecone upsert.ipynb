```python
import pinecone
from pinecone import Pinecone, ServerlessSpec
import os
```

```python
%load_ext dotenv
%dotenv
```

```python
pc = Pinecone(
    api_key=os.environ.get("PINECONE_API_KEY"),
    environment="gcp-starter"
)
```

```python
pc.list_indexes()
```

### Delete and Add new index

```python
index_name = "my-index"
dim = 3
metric = "cosine"
```

```python
if index_name in [index.name for index in pc.list_indexes()]:
    pc.delete_index(name=index_name)
```

```python
pc.list_indexes()
```

```python
pc.create_index(
    name=index_name,
    dimension=dim,
    metric=metric,
    spec=ServerlessSpec(
        cloud="aws",
        region="us-east-1"
    )
)
```

```python
pc.list_indexes()
```

## Upsert Data

```python
index = pc.Index(name=index_name)
```

```python
index.upsert([
    ("Dog", [4.0, 0.1, 1.0]),
    ("Cat", [4.0, 0.1, 1.0]),
    ("Chicken", [2.0, 2.1, 1.0]),
    ("Mantis", [6.0, 2.1, 3.0]),
    ("Elephant", [4.0, 0.1, 1.0]),
])
```
