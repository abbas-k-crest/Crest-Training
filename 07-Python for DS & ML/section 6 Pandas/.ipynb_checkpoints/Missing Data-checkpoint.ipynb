# Pandas Missing Data

```python
import numpy as np
import pandas as pd
```

```python
d = {'A':[1,2,np.nan], 'B':[4, np.nan, np.nan], 'C':[6,7,8]}
```

```python
d
```

```python
df = pd.DataFrame(d)
```

```python
df
```

```python
df.dropna()
```

```python
df.dropna(axis=1)
```

```python
df.dropna(thresh=2)
```

```python
df.fillna(value='Fill value')
```

```python
df
```

```python
df['A'].fillna(value=df['A'].mean(), inplace=True)
```

```python
df.fillna({'A': df['A'].mean()})
```

```python
df
```
