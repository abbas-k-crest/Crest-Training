```python
from groq import Groq
```

```python
client = Groq()
```

```python
messages = [{
    'role': 'user',
    'content': 'What is 2+5'
}]
```

```python
response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=messages
)
```

```python
response.choices[0].message.content
```
